```python
set1 = { 'A', 'B', 'C', 'D', 'E' }
set2 = { 1, 2, 3, 4, 5 }
set3 = set1.union(set2)
print(set3)
```

```python
set1 = { 1, 2, 3, 4, 5 }
set2 = { 4, 5, 6, 7, 8 }
set3 = set1.intersection(set2)
print(set3)
```
