```python
def readFile(filename):
    with open(filename, 'r') as file:
        content = file.read()
    return content

print(readFile('sample.txt'))
```

```python
def readFile(filename, n):
    with open(filename, 'r') as file:
        content = file.readlines()
    return content[:n]

n = int(input("Enter the number of lines to read: "))
print(''.join(readFile('sample.txt', n)))
```

```python
def appendToFile(filename, text):
    with open(filename, 'a') as file:
        file.write(text + '\n')

    with open(filename, 'r') as file:
        return file.read()

txt = input("Enter the text to append: ")
print(appendToFile('sample.txt', txt))
```

```python
def linesToList(filename):
    with open(filename, 'r') as file:
        content = file.readlines()
    return [line.strip() for line in content]

print(linesToList('sample.txt'))
```

```python
def longestWord(filename):
    with open(filename, 'r') as file:
        content = file.read()
        words = content.split()
    longest = max(words, key=len)
    return longest

print(longestWord('sample.txt'))
```

```python
def longestWord(filename, target):
    with open(filename, 'r') as file:
        content = file.read()
        words = content.split()
        count = words.count(target)
    return count

target = input("Enter the word to count: ")
print(f'The word "{target}" appears {longestWord("sample.txt", target)} times in the file.')
```
