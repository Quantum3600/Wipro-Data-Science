```python
def parseSecretMsg(filename):
    with open(filename, 'r') as file:
        content = file.read()
        words = content.split()
        lines = content.splitlines()
        hours = len(lines)
        if hours > 12:
            hours -= 12
            period = "PM"
        else:
            period = "AM"

        max = 0
        place = ""
        for word in words:
            count = words.count(word)
            if count > max:
                max = count
                place = 'Park Street' if word == 'park' else 'Oxford Street'

    print("Meet me at " + place + " at " + str(hours) + " " + period)

parseSecretMsg('projtxt.txt')
```
