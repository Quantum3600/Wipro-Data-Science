```python
def sumOfList(lst):
    total = 0
    for num in lst:
        total += num
    return total
lst = (8, 2, 3, 0, 7)
print("Sum of all elements in given list: ", sumOfList(lst))
```

```python
def reverseString(s):
    i = 0
    j = len(s)-1
    while i < j:
        s = list(s)
        s[i], s[j] = s[j], s[i]
        i += 1
        j -= 1
    return ''.join(s)

s = "Hello, World!"
print("Original string: ", s)
print("Reversed string: ", reverseString(s))
```

```python
def factorial(n):
    if n == 0: 
        return 1
    return n * factorial(n-1)

n = int(input("Enter a number to calculate its factorial: "))
print("Factorial of", n, "is", factorial(n))
```

```python
def noOfUpperLower(s):
    upper = 0
    lower = 0
    for c in s:
        if c.isupper():
            upper += 1
        elif c.islower():
            lower += 1
    return upper, lower

str = input("Enter a string to count uppercase and lowercase letters: ")
upper, lower = noOfUpperLower(str)
print("No. of Uppercase characters: ", upper)
print("No. of Lowercase characters: ", lower)
```

```python
def evenList(lst):
    even_numbers = []
    for num in lst:
        if num % 2 == 0:
            even_numbers.append(num)
    return even_numbers

lst = [1, 2, 3, 4, 5, 6, 7, 8, 9]
print("Even numbers in the list: ", evenList(lst))
```

```python
def isPrime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

n = int(input("Enter a number to check if it is prime: "))
if isPrime(n):
    print(n, "is a prime number.")
else:
    print(n, "is not a prime number.")
```
