```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv('Mall_Customers.csv')
num_data = df.select_dtypes(include='number')
plt.figure(figsize=(10, 8))
sns.heatmap(num_data.corr(), annot=True, cmap='coolwarm', linewidths=1)
```

```python
sns.pairplot(df)
```

```python
df.value_counts('Gender').plot(kind='bar', color=['pink', 'orange'])
```

```python
plt.figure(figsize=(14, 7)) 
sns.countplot(x='Age', data=df, width=0.8, gap=0.2)
```

```python
sns.scatterplot(
    x='Annual Income (k$)',
    y='Spending Score (1-100)',
    hue='Gender',
    data=df
)
```

```python
sns.pairplot(df, hue='Gender', palette=["red", "pink"])
```

```python
sns.FacetGrid(
    data=df,
    col='Gender',
    hue='Gender',
    height=3,
).map(
    sns.histplot,
    'Age',
    alpha=0.5
).add_legend()
```

```python
plt.figure(figsize=(16, 8))
sns.boxplot(x='Age', y='Annual Income (k$)', hue='Gender', data=df)
```
