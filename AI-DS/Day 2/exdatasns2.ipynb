```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv('Salary_Data.csv')
num_data = df.select_dtypes(include='number')
plt.figure(figsize=(10, 8))
sns.heatmap(num_data.corr(), annot=True, cmap='coolwarm', linewidths=1)
```

```python
sns.pairplot(df)
```

```python
df.value_counts('YearsExperience').plot(kind='bar', color=['pink', 'orange'])
```

```python
plt.figure(figsize=(14, 7)) 
sns.countplot(x='Salary', data=df, width=0.8, gap=0.2)
```

```python
sns.scatterplot(
    x='YearsExperience',
    y='Salary',
    data=df
)
```

```python
sns.pairplot(df, hue='YearsExperience', palette='Set1')
```

```python
sns.FacetGrid(
    data=df,
    col='Salary',
    hue='Salary',
    height=3,
).map(
    sns.histplot,
    'YearsExperience',
    alpha=0.5
).add_legend()
```

```python
plt.figure(figsize=(16, 8))
sns.boxplot(x='YearsExperience', y='Salary', data=df)
```
