```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
num_data = df.select_dtypes(include='number')
plt.figure(figsize=(10, 8))
sns.heatmap(num_data.corr(), annot=True, cmap='coolwarm', linewidths=1)
```

```python
sns.pairplot(df)
```

```python
df.value_counts('Purchased').plot(kind='bar', color=['Red', 'orange'])
```

```python
plt.figure(figsize=(14, 7)) 
sns.countplot(x='Age', data=df, width=0.8, gap=0.2)
```

```python
sns.scatterplot(
    x='EstimatedSalary',
    y='Age',
    hue='Purchased',
    data=df
)
```

```python
sns.pairplot(df, hue='Purchased')
```

```python
sns.FacetGrid(
    data=df,
    col='Gender',
    hue='Purchased',
    height=3,
).map(
    sns.histplot,
    'Age',
    alpha=0.5
).add_legend()
```

```python
plt.figure(figsize=(16, 8))
sns.boxplot(x='Age', y='EstimatedSalary', hue='Gender', data=df)
```
