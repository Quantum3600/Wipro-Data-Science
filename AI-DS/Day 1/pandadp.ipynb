```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('melb_data.csv')
features = df.drop(['Price'], axis=1)
labels = df['Price']
```

```python
numeric_features = features.select_dtypes(include=[np.number])
categorical_features = features.select_dtypes(exclude=[np.number])
categorical_features = categorical_features.fillna(categorical_features.mode().iloc[0], inplace=True)
numeric_features = numeric_features.fillna(numeric_features.mean(), inplace=True)
```

```python
categorical_dummies = pd.get_dummies(categorical_features)
updated_dataset = pd.concat([numeric_features, categorical_dummies], axis=1)
updated_dataset.head()
```
