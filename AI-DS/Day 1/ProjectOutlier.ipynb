```python
import numpy as np
import pandas as pd
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('datasetExample.csv')
df
```

```python
def outlierDetect(array):
    q1 = np.percentile(array, 25)
    q3 = np.percentile(array, 75)
    iqr = q3 - q1
    lf = q1 - (1.5 * iqr)
    uf = q3 + (1.5 * iqr)
    return lf, uf
numeric = df.select_dtypes(include=np.number)
for col in numeric:
    array = np.array(df[col])
    lf, uf = outlierDetect(array)
    outlier = array[(array < lf) | (array > uf)]
    print(f"For Column {col} outlier is {int(outlier[0])}")
```
