```python
import numpy as np
import pandas as pd
import sklearn as sk
import scipy as sp
```

```python
df = pd.read_csv('melb_data.csv')
features = df.drop(['Price'], axis=1)
label = df['Price']
```

```python
features['Date'] = features['Date'].astype('str')
numeric_features = features.select_dtypes(include=[np.number])
categorical_features = features.select_dtypes(exclude=[np.number])
```

```python
numeric_imputer = sk.impute.SimpleImputer(strategy='mean', missing_values=np.nan)
features[numeric_features.columns] = numeric_imputer.fit_transform(numeric_features)

categorical_imputer = sk.impute.SimpleImputer(strategy='most_frequent', missing_values=np.nan)
features[categorical_features.columns] = categorical_imputer.fit_transform(categorical_features)
```

```python
oh = sk.preprocessing.OneHotEncoder(sparse_output=True, handle_unknown='ignore')
categorical_features_encoded = oh.fit_transform(categorical_features)
output = sp.sparse.hstack([numeric_features, categorical_features_encoded])
```
