```python
import numpy as np
import pandas as pd
import sklearn as sk
import pickle
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
df.head()
```

```python
df.Purchased.value_counts()
```

```python
df['Gender'] = df['Gender'].map({'Male': 0, 'Female': 1})
```

```python
features = df.drop(['Purchased', 'User ID'], axis=1)
labels = df['Purchased']
```

```python
X_train, X_test, y_train, y_test = sk.model_selection.train_test_split(
    features, 
    labels, 
    test_size=0.2, 
    random_state=42
)
```

```python
scaler = sk.preprocessing.StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
model = sk.linear_model.LogisticRegression()
model.fit(X_train, y_train)
model.predict(X_test)
```

```python
model.score(X_train, y_train)
model.score(X_test, y_test)
```

```python
sk.metrics.confusion_matrix(y_test, model.predict(X_test))
```

```python
classification_report = sk.metrics.classification_report(y_test, model.predict(X_test))
print(classification_report)
```

```python
pickle.dump(model, open('social_model.pkl', 'wb'))
mymodel = pickle.load(open('social_model.pkl', 'rb'))
```
