```python
def lastdigit(num1, num2):
    if num1 % 10 == num2 % 10: return True 
    else: return False

num1 = int(input("give first number: "))
num2 = int(input("give second number: "))

print(lastdigit(num1, num2))
```

```python

import math

n = float(input("Enter a number: "))
root = math.sqrt(n)
flag = 0
for i in range(2, int(root)):
    if n % i == 0:
        flag = 1
        break
if flag == 1:
    print("Number not prime")
else:
    print("Number prime")
```

```python
summ = 0
n = int(input("Enter a number: "))
while n != 0:
    summ += n % 10
    n = n // 10
print(summ)
```

```python
rev = 0
n = int(input("Enter a number: "))
while n != 0:
    dig = n % 10
    rev = rev * 10 + dig
    n = int(n / 10)
print(rev)
```

```python
rev = 0
n = int(input("Enter a number: "))
fixed = n
while n != 0:
    dig = n % 10
    rev = rev * 10 + dig
    n = int(n / 10)
if rev == fixed:
    print("number is palindrome")
else:
    print("number is not palindrome")
```

```python
n = int(input("Enter a number: "))
if n < 2:
    print("Not prime")
else:
    for i in range(2, n):
        if n % i == 0:
            print("Not prime")
            break
    else:
        print("Prime")
```
