```python
import re

lst = ['789', '123', '004', '729']
pattern = r'^[0-7]+$'
for string in lst:
    m = re.findall(pattern, string)
    if m: print(string)
```

```python
pattern = r'(\w+)@(\w+).((..?)|(..?).(..?))'
string = """zuck@facebook.co.in
sunder33@google.com
jeff42@amazon.com"""
m = re.findall(pattern, string)
for i in m:
    print(f"User ID: {i[0]}")
    print(f"Domain Name: {i[1]}")
    print(f"Suffix: {i[2]}")
```

```python
pattern = r'[a-zA-Z]+'
string = """A, very very; irregular_sentence"""
m = re.findall(pattern, string)
for i in m:
    print(i)
```

```python
string = '''Good advice! RT @TheNextWeb: What I would do differently if I was learning to code today http://t.co/lbwej0pxOd cc: @garybernhardt #rstats'''
string = re.sub(r'https?://\S+', '', string)
pattern = r'\w+'
m = re.findall(pattern, string)
print(m)
```

```python
import requests
r = requests.get("https://raw.githubusercontent.com/selva86/datasets/master/sample.html")  # html text is contained here
pattern = r'<[^>]+>'
text = re.sub(pattern, ' ', r.text)
text = re.sub(r'\s+', ' ', text).strip()
print(text)
```
