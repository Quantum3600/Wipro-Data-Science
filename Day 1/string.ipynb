```python
str = "Swarnali Amar Sathe Jhogra Kore"
countup = 0
countlow = 0
for ch in str:
    if ch.isupper():
        countup += 1
    elif ch.islower():
        countlow += 1
print(f"Uppercase letters: {countup}")
print(f"Lowercase letters: {countlow}")
```

```python
str = "madam"
i = 0
j = len(str) - 1
while i < j:
    if str[i] != str[j]:
        print("Not a palindrome")
        break
    i += 1
    j -= 1
else:
    print("Is a palindrome")
```
