```python
dict = {"Alice" : "Is Afraid of the Dark", "Bob" : "Loves to Play Football", "Charlie" : "Enjoys Reading Books", "David" : "Is Afraid of Heights"}
for pair in dict.items():
    print(f"{pair[0]}: {pair[1]}")
dict["Bob"] = "Loves to Play Basketball"
dict["Eve"] = "Is Afraid of Spiders"
print("\nUpdated Dictionary:")
for pair in dict.items():
    print(f"{pair[0]}: {pair[1]}")
```

```python
scores = [85, 92, 78, 96, 88]
winner = 0
runner_up = 0
for score in scores:
    if score > winner:
        runner_up = winner
        winner = score
    elif score > runner_up and score != winner:
        runner_up = score
print(f"Runner-up: {runner_up}")
```

```python
def calc_avg(lst):
    if not lst:
        return 0
    return sum(lst) / len(lst)

sample = {"Krishna" : [67,68,69], "Arjun" :[70,98,63], "Malika" : [52,56,60] }
map = {}
for key in sample:
    map[key] = calc_avg(sample[key])
n = input("Enter the name of the student to get the average marks: ")
print(f"Average marks of {n}: {map.get(n, 'Student not found')}")
```
